##### **3. Particle Swarm Optimization**

![PSO graph.png](<attachment:PSO graph.png>)

These SHAP value plots represents the average impact of different genes on the output of a predictive model for hepatocellular carcinoma (HCC) staging. The x-axis shows the mean absolute SHAP value, which indicates the average contribution of each gene to the model's predictions, whereas the y-axis lists the genes used as features. Higher SHAP values mean that the gene has a stronger impact on the model’s decision. The colored segments within each bar represent the impact of the gene in different HCC stages, such as EHCC being associated with blue or PHCC with purple.

DNAJB14 can be observed to be the most influential gene in all three models, with a consistently high SHAP value across different HCC stages. DNAJB14's contribution is primarily towards predicting the EHCC and PHCC classes, suggesting its importance in distinguishing these stages. Other genes show a much more balanced influence across different stages. The relative importance of genes differs between models, because different optimization approaches (Genetic Algorithm, PSO, and Ant Colony Optimization) select different features with varying levels of contribution. DNAJB14 appears as the top feature in all models. ZNF415 appears in the first and second graphs, whereas RIMBP2 and HGF appear in the first and third graphs.

The third model (Ant Colony Optimization-based) produces significantly higher SHAP values, suggesting that it might have identified genes with a stronger contribution to classification. The first two models (Genetic Algorithm and PSO) have lower but more evenly distributed SHAP values across genes, indicating that they may rely on a more balanced set of features.
